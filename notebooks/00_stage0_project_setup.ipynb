```python
def find_project_root():
    """
    Find project root.
    
    If notebook is inside notebooks/, project root is parent.
    Otherwise current directory is project root.
    """
    current = Path.cwd().resolve()
    
    if current.name.lower() == "notebooks":
        return current.parent
    
    return current


PROJECT_ROOT = find_project_root()

print("Current working directory:")
print(Path.cwd())

print("\nProject root:")
print(PROJECT_ROOT)
```

```python
def count_files(folder):
    folder = Path(folder)
    
    if not folder.exists():
        return 0
    
    allowed_extensions = {".csv", ".xlsx", ".xls", ".txt", ".dat"}
    
    files = [
        p for p in folder.rglob("*")
        if p.is_file() and p.suffix.lower() in allowed_extensions
    ]
    
    return len(files)


fc1_file_count = count_files(FC1_RAW_DIR)
fc2_file_count = count_files(FC2_RAW_DIR)
total_raw_files = count_files(RAW_DIR)

print("FC1 raw files:", fc1_file_count)
print("FC2 raw files:", fc2_file_count)
print("Total raw files:", total_raw_files)
```

```python
def list_first_files(folder, n=10):
    folder = Path(folder)
    
    if not folder.exists():
        print(f"Folder does not exist: {folder}")
        return
    
    files = sorted([p for p in folder.rglob("*") if p.is_file()])
    
    for file in files[:n]:
        print(file.relative_to(PROJECT_ROOT))


print("First FC1 files:")
list_first_files(FC1_RAW_DIR, n=10)

print("\nFirst FC2 files:")
list_first_files(FC2_RAW_DIR, n=10)
```
